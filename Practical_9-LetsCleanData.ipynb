```python
def zipper_file(filename, strings):
    with open(filename, "w") as file:
        for i in range(0, len(strings), 2):
            first = strings[i]
            second = strings[i + 1]

            result = ""

            for j in range(max(len(first), len(second))):
                if j < len(first):
                    result += first[j]

                if j < len(second):
                    result += second[j]

            file.write(result + "\n")
```

```python
filename = "zipped_output.txt"

strings = [
    "ABC", "123",
    "HELLO", "WORLD",
    "PYTHON", "CODING"
]

zipper_file(filename, strings)

print("File created successfully!")
```

```python
with open(filename, "r") as file:
    print(file.read())
```

```python
def reformat_date(dates):

    months = {
        "01": "January",
        "02": "February",
        "03": "March",
        "04": "April",
        "05": "May",
        "06": "June",
        "07": "July",
        "08": "August",
        "09": "September",
        "10": "October",
        "11": "November",
        "12": "December"
    }

    result = []

    for date in dates:
        month, day, year = date.split("/")

        month_name = months[month]

        sentence = f"Office Hours on {month_name} {day}, {year}"

        result.append(sentence)

    return result
```

```python
dates = [
    "01/15/2026",
    "02/05/2026",
    "03/21/2026",
    "06/10/2026",
    "12/25/2026"
]

formatted_dates = reformat_date(dates)

for date in formatted_dates:
    print(date)
```
