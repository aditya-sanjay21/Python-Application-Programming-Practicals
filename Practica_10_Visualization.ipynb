```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Set visual theme
sns.set_theme(style="whitegrid")

# Load the built-in Iris dataset
iris = sns.load_dataset("iris")

# Display first 5 rows to verify
iris.head()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=iris, x="petal_length", y="petal_width", hue="species", style="species", s=80)
plt.title("Scatter Plot: Petal Length vs Petal Width by Species")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()
```

```python
Analysis:

There is a strong positive linear relationship between petal length and petal width across all species: as petal length increases, petal width also increases.

Setosa is clearly separated from the other two species, clustering tightly in the bottom-left with the smallest petals (length < 2 cm, width < 0.7 cm).

Versicolor and Virginica show larger petal dimensions and exhibit a slight overlap around petal lengths of 4.5 to 5.1 cm, though Virginica generally has the largest petals.
```

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data=iris, x="sepal_length", hue="species", fill=True, common_norm=False, alpha=0.5)
plt.title("Density Plot: Distribution of Sepal Length by Species")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Density")
plt.show()
```

```python
Analysis:

The Kernel Density Estimate (KDE) plot illustrates the probability distribution of sepal lengths for each species.

Setosa has a narrow, tall peak centered around 5.0 cm, indicating low variance and tightly clustered sepal lengths.

Versicolor (peaking around 5.9 cm) and Virginica (peaking around 6.5 cm) have wider, flatter curves with significant overlap, showing greater spread in their sepal lengths compared to Setosa
```

```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=iris, x="species", y="sepal_width", hue="species", palette="muted", legend=False)
plt.title("Violin Plot: Sepal Width Distribution across Species")
plt.xlabel("Species")
plt.ylabel("Sepal Width (cm)")
plt.show()
```

```python
Analysis:

The violin plot combines a box plot (showing median and interquartile range) with a density curve on each side.

Setosa has the highest median sepal width (~3.4 cm) and a relatively symmetrical distribution.

Versicolor has the lowest median sepal width (~2.8 cm), while Virginica sits slightly higher (~3.0 cm) and displays a few visible outliers at both ends (near 2.2 cm and 3.8 cm).
```

```python
plt.figure(figsize=(8, 6))
# Compute correlation matrix for numerical columns only
corr_matrix = iris.drop(columns=["species"]).corr()

sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Heatmap of Iris Numerical Features")
plt.show()
```

```python
Analysis (In Context of Correlation):Strong Positive Correlation: petal_length and petal_width have the highest positive correlation ($r = 0.96$), meaning they increase together almost linearly. sepal_length also shows strong positive correlations with petal_length ($r = 0.87$) and petal_width ($r = 0.82$).Weak/Negative Correlation: sepal_width has a weak negative correlation with petal_length ($r = -0.43$) and petal_width ($r = -0.37$), and almost no linear correlation with sepal_length ($r = -0.12$). This indicates that sepal width varies mostly independently of the other features.
```

```python
joint = sns.jointplot(data=iris, x="sepal_length", y="petal_length", hue="species", kind="scatter", height=7)
joint.fig.suptitle("Joint Plot: Sepal Length vs Petal Length", y=1.02)
plt.show()
```

```python
Analysis:

The joint plot displays both the bivariate scatter relationship in the center and the univariate marginal density distributions along the top and right axes.

For Versicolor and Virginica, there is a clear upward trend showing that flowers with longer sepals also tend to have longer petals.

Setosa, however, shows almost no correlation between sepal length and petal length—its petal length stays nearly constant (~1.4–1.5 cm) even as its sepal length varies from 4.3 to 5.8 cm
```

```python
pair = sns.pairplot(iris, hue="species", markers=["o", "s", "D"])
pair.fig.suptitle("Pair Plot of All Iris Features by Species", y=1.02)
plt.show()
```

```python
Analysis:

The pair plot provides a complete matrix of scatter plots for every pair of features alongside diagonal KDE plots for individual feature distributions.

Feature Separability: Plots involving petal_length and petal_width provide the cleanest separation between all three species, making them the most useful features for classifying iris species.

Conversely, the scatter plot between sepal_length and sepal_width shows heavy overlap between Versicolor and Virginica, making those two species difficult to distinguish using sepal dimensions alone.
```
