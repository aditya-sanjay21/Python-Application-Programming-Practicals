```python
data = {
    'Customer_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Aman', 'Rahul', 'Priya', 'Neha', 'Rohan',
             'Anjali', 'Karan', 'Simran', 'Vikas', 'Pooja'],
    'Age': [22, 25, 31, 28, 35, 24, 29, 32, 26, 30],
    'Gender': ['Male', 'Male', 'Female', 'Female', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],
    'Purchase_Amount': [2000, 3500, 4500, 3000, 5000,
                        2500, 4000, 5500, 3200, 4800]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.head()
```

```python
df.describe()
```

```python
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())

df['Purchase_Amount'] = df['Purchase_Amount'].fillna(
    df['Purchase_Amount'].mean()
)

print(df.isnull().sum())
```

```python
customers = df[df['Age'] > 30]

print(customers)
```

```python
gender_total = df.groupby('Gender')['Purchase_Amount'].sum()

print(gender_total)
```

```python
selected_data = df[['Customer_ID', 'Name', 'Purchase_Amount']]

print(selected_data)
```

```python
df['Discounted_Price'] = df['Purchase_Amount'] * 0.90

print(df)
```

```python
average_purchase = df.groupby('Gender')['Purchase_Amount'].mean()

print(average_purchase)
```

```python
sorted_df = df.sort_values(by='Purchase_Amount', ascending=False)

print(sorted_df)
```
